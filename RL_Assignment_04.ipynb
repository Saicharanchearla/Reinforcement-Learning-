```python
import math

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Part B – Initialize knowledge tables
N = {machine: 0 for machine in machines}       # Participation count
R = {machine: [] for machine in machines}      # Rewards received
Q = {machine: 0.0 for machine in machines}     # Estimated Q-values
UCB = {machine: float('inf') for machine in machines}  # UCB values

total_reward = 0

print("=== Upper Confidence Bound (UCB) Multi-Armed Bandit ===\n")

# Part C, D and E – UCB selection, reward collection and Q update
for play in range(1, N_play + 1):
    # Calculate UCB values
    for machine in machines:
        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    # Choose machine with highest UCB
    selected_machine = max(machines, key=lambda x: UCB[x])

    print(f"Play {play}")
    print(f"Selected Machine: {selected_machine}")

    # Prompt user for the reward input
    reward_input = input("Enter reward: ")
    try:
        reward = float(reward_input)
    except ValueError:
        reward = 0.0

    print(f"Reward obtained: {reward}")

    # Update stats
    N[selected_machine] += 1
    R[selected_machine].append(reward)
    total_reward += reward

    # Update estimated Q-value
    Q[selected_machine] = sum(R[selected_machine]) / N[selected_machine]
    print(f"Updated Q({selected_machine}) = {Q[selected_machine]:.2f}\n")

# Final results display
print("=" * 60)
print("FINAL RESULTS")
print("=" * 60 + "\n")
print("Machine\tN\tRewards\t\tUCB\t\tQ-value")
for machine in machines:
    # Recalculate final UCB after the last round
    if N[machine] == 0:
        final_ucb = float('inf')
    else:
        final_ucb = Q[machine] + math.sqrt((2 * math.log(N_play)) / N[machine])
    print(f"{machine}\t{N[machine]}\t{R[machine]}\t{final_ucb:.2f}\t\t{Q[machine]:.2f}")

best_machine = max(machines, key=lambda x: Q[x])
print(f"\nTotal Reward Received = {total_reward}")
print(f"Best Machine = {best_machine}")
print(f"Highest Estimated Q-value = {Q[best_machine]:.2f}")
```
